```python
import pandas as pd
import xml.etree.ElementTree as ET

# Parse the XML file
tree = ET.parse('export.xml')
root = tree.getroot()

# Extract Records
data = []

for record in root.findall('Record'):
    record_dict = {
        'type': record.attrib.get('type'),
        'sourceName': record.attrib.get('sourceName'),
        'sourceVersion': record.attrib.get('sourceVersion'),
        'device': record.attrib.get('device'),
        'unit': record.attrib.get('unit'),
        'creationDate': record.attrib.get('creationDate'),
        'startDate': record.attrib.get('startDate'),
        'endDate': record.attrib.get('endDate'),
        'value': record.attrib.get('value'),
    }
    data.append(record_dict)

# Create a DataFrame
df = pd.DataFrame(data)
df['startDate'] = pd.to_datetime(df['startDate'])
df['endDate'] = pd.to_datetime(df['endDate'])
df['value'] = pd.to_numeric(df['value'], errors='coerce')
```

```python
# Export DataFrame to CSV
df.to_csv('fitness_data.csv', index=False)
```

```python
# Filter for different types
heart_rate_df = df[df['type'] == 'HKQuantityTypeIdentifierHeartRate']
step_count_df = df[df['type'] == 'HKQuantityTypeIdentifierStepCount']
distance_df = df[df['type'] == 'HKQuantityTypeIdentifierDistanceWalkingRunning']
active_energy_df = df[df['type'] == 'HKQuantityTypeIdentifierActiveEnergyBurned']

# Display the first few rows of each DataFrame
print("Heart Rate Data:")
print(heart_rate_df.head())

print("\nStep Count Data:")
print(step_count_df.head())

print("\nDistance Data:")
print(distance_df.head())

print("\nActive Energy Data:")
print(active_energy_df.head())
```
